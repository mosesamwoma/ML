```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
print(sns.get_dataset_names())
```

```python
tips=sns.load_dataset('tips')
titanic=sns.load_dataset('titanic')
penguins=sns.load_dataset('penguins')
```

```python
tips.head()
```

```python
sns.set_theme()
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', size='size', style='time')
```

```python
sns.histplot(data=tips, x='total_bill', bins=20, kde=True)
```

```python
sns.boxplot(data=tips,x='day',y='total_bill',hue='sex', dodge=True,gap=0.15)
```

```python
penguins.head()
```

```python
sns.violinplot(data=penguins, x='species', y='body_mass_g', hue='sex')
```

```python
titanic.head()
```

```python
sns.countplot(data=titanic, x='class', hue='sex')
```

```python
sns.barplot(data=titanic, x='class', y='survived', hue='class')
```

```python
sns.jointplot(data=penguins, x='bill_length_mm', y='bill_depth_mm',hue='sex', kind='scatter')
plt.show()
```

```python
numeric_titanic = titanic.select_dtypes(include='number')
sns.heatmap(data=numeric_titanic.corr(),annot=True,cmap='coolwarm',center=0)
```

```python
sns.pairplot(data=penguins, hue="species")
```

```python
numeric_titanic = titanic[["age", "fare", "survived", "pclass", "sex"]].dropna()
g = sns.PairGrid(data=numeric_titanic, hue="sex", corner=True, palette="Set2",height=2.5)
g.map_lower(sns.kdeplot, hue=None, levels=5, color=".2")
g.map_lower(sns.scatterplot, marker="+")
g.map_diag(sns.histplot, element="step", linewidth=0, kde=True)
g.add_legend(frameon=True)
g.legend.set_bbox_to_anchor((0.9, 0.9)) 
for ax in g.axes[:, 0]:
    if ax is not None and ax.get_ylabel() in ["survived", "pclass"]:
        ax.set_yticks([1, 2, 3] if ax.get_ylabel() == "pclass" else [0, 1])
plt.show()
```
